<h2 style='font-size:30px'> 11.1 Sequences</h2>
<div> 
    <ul style='font-size:20px'> 
        <li>
            Uma sequência seria uma ordenação de números. $a_{n}$ corresponderia ao $n$-ésimo valor da sequência.
        </li>
        <li>
            Nessa seção, todas as sequências serão <i> infinitas</i>. Podemos defini-las de diversas maneiras, como $\{a_{1},a_{2},\ldots\}$,$\{a_{n}\}$,$\{a_{n}\}_{n=1}^{\infty}$. Repare que $n$ sempre será um positivo maior ou igual a 1!
        </li>
    </ul>
</div>

<h2 style='font-size:30px'> 11.2 Series</h2>
<div> 
    <ul style='font-size:20px'> 
        <li>
            Seja $\{a_{n}\}$ uma sequência infinita de números, entendemos $\displaystyle\sum_{n=1}^{\infty}{a_{n}}$ como uma série.
        </li>
        <li>
            Podemos, nessa mesma sequência, querer fazer uma soma até um $n$-ésimo elemento arbitrário. Nesse caso, chamamos essa operação de <strong> partial sum</strong>.
        </li>
    </ul>
</div>

<figure>
    <center style='font-size:20px;margin-top:20px'>
        $\displaystyle s_{n}=\sum_{i=1}^{n}{a_{i}}$
        <figcaption style='font-size:15px'> 
            Definição de partial sum.
        </figcaption>
    </center>
</figure>

<h3 style='font-size:30px;font-style:italic'> 
    Convergência/Divergência de Séries
</h3>
<div> 
    <ul style='font-size:20px'>
        <li>
            Seja a uma série $\displaystyle\sum{a_{n}}$ e $\{s_{n}\}$ a sequência de suas partial sums, podemos dizer que ela é <strong>convergente</strong>, se essa sequência também o for e $\displaystyle\lim_{n\to\infty}{s_{n}}=s$ e sej um número real.
        </li>
        <li>
            Caso o contrário, ela será <strong>divergente</strong>.
        </li>
     </ul>
</div>
